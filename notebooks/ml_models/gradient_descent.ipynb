```python
import numpy as np

# 100 samples, 3 features
X = np.random.randn(100, 3) 

# The true weights we want your algorithm to discover
true_W = np.array([[1.5], [-2.0], [3.0]]) 

# Target variable y (keeping it simple with zero noise)
y = X.dot(true_W) 

# Initialize your model's weights randomly
W = np.random.randn(3, 1) 

alpha = 0.1
epochs = 100
N = X.shape[0]
```

```python
for epoch in range(epochs):

    y_pred = X @ W
    error = y_pred - y
    gradient = (2/N)* X.T @ error
    W = W - alpha * gradient

print("Learned weights:\n", W)
```
