```python
import numpy as np

class LinearSVM:
    def __init__(self, learning_rate=0.001, C=1.0, n_iters=1000):
        self.learning_rate = learning_rate
        self.C = C
        self.n_iters = n_iters
        self.w = None
        self.b = None

    def _init_weights(self,X):
        self.w = np.zeros(X.shape[1], dtype=float)
        self.b = 0.0

    def decision_function(self,X):
        return X@self.w + self.b

    def fit(self,X,y):
        self._init_weights(X)
        np.where(y == 0, -1, 1)
        for epoch in range(self.n_iters):
            margin_values = y * self.decision_function(X)
            grad_w = self.w.copy()
            grad_b = 0.0

            misclassified_X = X[margin_values < 1]
            misclassified_Y = y[margin_values < 1]

            grad_w = grad_w - np.sum(self.C * misclassified_Y[:,np.newaxis] * misclassified_X, axis=0)
            grad_b = grad_b - np.sum(self.C * misclassified_Y)
        #    for i in range (X.shape[0]):
         #       if margin_values[i]<1:
          #          grad_w = grad_w - self.C * y[i] *X[i]
           #         grad_b = grad_b - self.C * y[i]

            self.w = self.w - self.learning_rate * grad_w
            self.b = self.b - self.learning_rate * grad_b

    def predict(self, X):
        return np.sign(self.decision_function(X))
```

```python
from sklearn.datasets import make_blobs

# 1. Create a simple linearly separable dataset
X_train, y_train = make_blobs(n_samples=100, centers=2, random_state=42, cluster_std=1.0)

# 2. Scikit-learn outputs labels as 0 and 1. We MUST convert 0 to -1 for our math to work!
y_train = np.where(y_train == 0, -1, 1)

# 3. Instantiate and fit your model
svm = LinearSVM(learning_rate=0.001, C=1.0, n_iters=1000)
svm.fit(X_train, y_train)

# 4. Check accuracy
predictions = svm.predict(X_train)
accuracy = np.mean(predictions == y_train)
print(f"Accuracy: {accuracy * 100}%")
```
