```python
import numpy as np

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

# 100 samples, 3 features
X = np.random.randn(100, 3) 

# The "true" weights driving the underlying pattern
true_W = np.array([[1.5], [-2.0], [3.0]]) 

# Generate raw continuous values, then threshold them to create binary labels (0 or 1)
z_true = X.dot(true_W)
y = (z_true > 0).astype(int) 

# Initialize random weights
W = np.random.randn(3, 1) 

alpha = 0.1
epochs = 100
N = X.shape[0]

for epoch in range(epochs):
    y_pred = sigmoid(X @ W)
    error = y_pred - y
    gradient = X.T @ error / N
    W= W-alpha * gradient

# Convert probabilities to 0 or 1 based on a 0.5 threshold
final_predictions = (sigmoid(X @ W) >= 0.5).astype(int)

# Calculate accuracy
accuracy = np.mean(final_predictions == y) * 100
print(f"Model Accuracy: {accuracy}%")
print(W)
```
